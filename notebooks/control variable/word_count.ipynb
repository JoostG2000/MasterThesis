```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import os
import wrds

def find_project_root(start_path, marker="data"):
    path = Path(start_path).resolve()
    for parent in [path] + list(path.parents):
        if (parent / marker).exists():
            return parent
    raise RuntimeError("Project root not found")

PROJECT_ROOT = find_project_root(Path.cwd())
os.chdir(PROJECT_ROOT)

print("Project root:", PROJECT_ROOT)

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
DATA_SAMPLES =  PROJECT_ROOT / "data" / "samples"
DATA_OUTPUT =  PROJECT_ROOT / "data" / "output"
```

```python
data = pd.read_csv(DATA_RAW / "transcripts_final.csv") #enter filename here
```

```python
word_count = data[[ 'transcriptid', 'transcript_text', 'companyid', 'companyname',
       'mostimportantdateutc']].copy()
```

```python
import re
import numpy as np

def clean_text(text):
    text = re.sub(r"\bAnalysts:\b", "", text)
    text = re.sub(r"\bExecutives:\b", "", text)
    return text

word_count["clean_text"] = (
    word_count["transcript_text"]
    .astype(str)
    .apply(clean_text)
)

word_count["word_count"] = (
    word_count["clean_text"]
    .str.split()
    .str.len()
)

word_count["log_word_count"] = np.log(
    word_count["word_count"]
)
```

```python
word_count.drop(columns=["clean_text"], inplace=True)
```

```python
word_count.to_csv(DATA_PROCESSED / "word_count.csv", index=False)
```
