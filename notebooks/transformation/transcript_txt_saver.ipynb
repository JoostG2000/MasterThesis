```python
from pathlib import Path
import pandas as pd
import os
```

```python
def find_project_root(start_path, marker="data"):
    path = Path(start_path).resolve()
    for parent in [path] + list(path.parents):
        if (parent / marker).exists():
            return parent
    raise RuntimeError("Project root not found")

PROJECT_ROOT = find_project_root(Path.cwd())
os.chdir(PROJECT_ROOT)

print("Project root:", PROJECT_ROOT)
```

```python
DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
```

```python
os.makedirs("transcripts_qna", exist_ok=True)

tids = (2537353,2089204,2723571,2431130,2788125,2934173,2785447,3208500,3023856,3014724)
docs_df_all = pd.read_csv(DATA_RAW/'transcripts_final.csv')
```

```python
docs_df_all.columns
```

```python
docs_df_sample = docs_df_all.loc[docs_df_all["transcriptid"].isin(tids)]
```

```python
for _, row in docs_df_sample.iterrows():

    companyid = row["companyid"]
    date = row["mostimportantdateutc"]
    transcriptid = row["transcriptid"]
    text = row["transcript_text"]

    filename = f"transcripts_qna/{companyid}_{date}_{transcriptid}.txt"

    with open(filename, "w", encoding="utf-8") as f:
        f.write(text)
```
